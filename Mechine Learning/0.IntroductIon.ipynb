
### 모델 검증(Model Validation)이란?
- 대부분의 경우 모델의 품질을 측정하는 주요 기준은 **예측 정확도(Predictive Accuracy)** 입니다. 즉, 모델의 예측값이 실제 결과와 얼마나 가까운지를 확인하는 것입니다.
- 훈련 데이터(training data)를 사용하여 예측한 값을 훈련 데이터의 실제 값과 비교하여 모델을 평가합니다.

</br>

### 평균 절대 오차(Mean Absolute Error, MAE)린?

1. 오차(Error)란?
- 예측값과 실제 값의 차이를 **오차(Error)**라고 합니다.

2. 평균 절대 오차(MAE) 계산 방법
- MAE는 오차의 절대값을 평균 내는 방식으로 모델의 품질을 측정합니다.
- **오차 = 실제값 - 예측값**
- 모든 오차의 평균을 구합니다.
- 결과적으로 **"평균적으로 우리의 예측값이 실제 값과 X 정도 차이가 난다."**라고 해석할 수 있습니다.

</br>

### MAE 계산 예제
- 모델을 만든 후, 평균 절대 오차 계산

```python
from sklearn.metrics import mean_absolute_error


# 예측값 생성
predicted_home_prices = iowa_model.predict(X)

# 평균 절대 오차 계산
mean_absolute_error(y,predicted_home_prices)
```

### "훈련 데이터로 평가하는 방식"의 문제점
- 위 계산한 MAE값은 훈련 데이터를 사용하여 평가한 값입니다. 이런 방식의 평가를 **In-Sample Score (훈련 데이터 기반 점수)** 라고 합니다. 이런 방식의 평가는 아래와 같은 현실적인 문제를 초래할 수 있습니다.
</br>

### 녹색 문(Green Door) 문제
- 실제 부동산 시장에서는 문 색깔이 집값에 영향을 주지 않는다고 가정
- 그런데 우리가 모델을 훈련할 때 사용한 데이터에서는 녹색 문을 가지 집들이 매우 비쌌습니다.
- 모델은 데이터를 학습하면서 "녹색 문이 있으면 집값이 비싸다" 라는 패턴을 학습하게 됩니다
- 따라서, 새로운 데이터에서 녹색 문이 있는 집은 무조건 비싸게 예측하게 됩니다.
- 그러나 실제 부동산 시장에서는 이 패턴이 적용되지 않으므로, 모델은 새로운 데이터에서 매우 부정확한 결과를 내놓을 것입니다.

즉, 훈련 데이터에서 얻은 패턴이 새로운 데이터에서 동일하게 적용되지 않을 경우, 모델의 성능이 크게 저하됩니다.  
이 문제를 해결하기 위해, 모델을 훈련에 사용되지 않는 데이터로 평가해야 합니다. 이를 **검증 데이터(validation data)** 라고 합니다.
</br>

### 모델 검증을 위한 데이터 분할 (train_test_split)
- Scitit-learn의 train_test_split 함수를 사용하여 데이터를 훈련 데이터(training data)와 검증 데이터(validation data)로 나눌 수 있습니다.

```python
from sklearn.model_selection import train_test_split

#데이터를 훈련용과 검증용으로 분할
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=0)


# 모델 정의
melbourne_model = DecisionTreeRegressor()

# 모델 학습
melbourne_model.fit(train_X, train_y)

# 모델 예측
val_predictions = melbourne_model.predict(val_X)

# 평균 절대 오차 계산
print(mean_absolute_error(val_y, val_predictions))
```

### 결과 분석 및 문제점 발견
- 훈련 데이터에서 측성한 MAE와 검증 데이터에서 측정한 MAE에서 큰 오차 발생

</br>

### 이 문제는 왜 발생했을까?
- 모델이 훈련 데이터의 패턴을 **과도하게 학습(Overfitting, 과적합)**했기 때문입니다.  
- 실제 부동산 시장에서는 적용되지 않는 불필요한 패턴까지 학습했기 때문에, 새로운 데이터에서 성능이 크게 떨어졌습니다.

</br>

### 결론 및 다음 단계
- 모델의 성능을 올바르게 평가하려면 훈련 데이터가 아닌 새로운 검증 데이터(validation data)로 평가해야 합니다.
- 현재 모델은 새로운 데이터에서 예측 오류가 너무 크기 때문에, 실무에서 사용하기 어렵습니다.
- 모델을 개선하려면 더 나은 특징(feature)을 찾거나, 다른 유형의 모델을 실험해봐야 합니다.

### 다양한 모델 실험하기 (Experimenting With Different Models)

결정 트리의 깊이 (Tree Depth)
- Scikit-learn의 결정 트리 모델에는 다양한 옵션이 있습니다. 이 중에서 **가장 중요한 옵션은 트릐의 깊이**를 결정하는 것입니다.
- 트리의 깊이는 예측을 내리기 전까지 데이터를 얼마나 많이 분할하는지를 의미합니다.

</br>

📌 트리의 깊이가 깊어지면 어떻게 될까?
- 장점 : 각 리프 토느의 예측 값은 해당 노드에 있는 실제 값과 매우 가까워진다.
- 단점 : 새로운 데이터에 대해서는 예측이 부정확해질 가능성이 크다.  
즉, 훈련 데이터에서는 완벽한 성능을 내지만 **새로운 데이터에서는 성능이 낮아지는 "과정합(Overfitting)"** 이 발생한다.

</br>

📌 반대로, 트리의 깊이가 너무 얕으면?
- 데이터가 충분히 세분화되지 않아 여러 유형의 데이터를 같은 그룹으로 예측하게 된다.  
이렇게 되면 예측값이 실제값과 크게 차이가 나는 **"과소적합(Underfitting)"** 이 발생하게 된다.

</br>

📌 과적합 vs 과소적합 (Overfitting vs Underfitting)
- 과적합(Overfitting): 훈련 데이터에는 잘 맞지만, 새로운 데이터에는 성능이 낮음
- 과소적합(Underfitting): 훈련 데이터에서도 성능이 낮고, 검증 데이터에서도 성능이 낮음  
우리는 훈련 데이터와 검증 데이터 모두에서 성능이 좋은 "적절한 깊이"를 찾아야 합니다!


### 최적의 트리 깊이 찾기 (Finding the Best Tree Depth)
- 트리 깊이를 조절하는 여러 방법이 있지만, **max_left_nodes**를 조정하는 것이 가장 직관적인 방법 중 하나.
- max_leaf_nodes ↑ = 과적합으로 이동
- max_leaf_nodes ↓ = 과소적합으로 이동

</br>


### 🔍 최적의 max_leaf_nodes 찾기
- 아래와 같은 함수(get_mae) 를 만들어서 max_leaf_nodes 값에 대한 MAE를 비교할 수 있습니다.

```python
from sklearn.metrics import mean_absolute_error
from sklearn.tree import DecisionTreeRegressor

# MAE 계산 함수
def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return mae


# 서로 다른 max_leaf_nodes 값을 비교
for max_leaf_nodes in [5, 50, 500, 5000]:
    my_mae = get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y)
    print("Max leaf nodes: %d  \t\t Mean Absolute Error:  %d" %(max_leaf_nodes, my_mae))
```

- max_leaf_nodes=5 → 과소적합 (Underfitting)
- max_leaf_nodes=5000 → 과적합 (Overfitting)
- max_leaf_nodes=50 → 가장 낮은 MAE 값 → 최적의 값! ✅


</br>

### 더 나은 예측 수행하기 - 트리의 크기 최적화



```python
# 데이터 로드에 사용한 코드
import pandas as pd
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor

# 파일 경로
iowa_file_path = 'data/train.csv'

# 데이터 읽기
home_data = pd.read_csv(iowa_file_path)

# 타겟 변수 (y) 생성
y = home_data.SalePrice

# 특성 선택 (X)
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']
X = home_data[features]

# 데이터 분할 (훈련 데이터와 검증 데이터)
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

# 모델 지정
iowa_model = DecisionTreeRegressor(random_state=1)

# 모델 학습
iowa_model.fit(train_X, train_y)

# 검증 데이터로 예측 수행 및 MAE 계산
val_predictions = iowa_model.predict(val_X)
val_mae = mean_absolute_error(val_predictions, val_y)

# 결과 출력
print("Validation MAE: {:,.0f}".format(val_mae))
```

### [1단계] : 다양한 트리 크기 비교하기 (Step 1: Compare Different Tree Sizes)

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error

def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return mae
```

### 🔍 최적의 트리 크기 찾기
다음 값에 대해 get_mae 함수를 실행하여 가장 정확한 모델을 찾습니다.

```python
# 후보 max_leaf_nodes 값
candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500]

# 각 트리 크기에 대한 MAE 계산
scores = {leaf_size: get_mae(leaf_size, train_X, val_X, train_y, val_y) for leaf_size in candidate_max_leaf_nodes}

# MAE가 가장 작은 (최적의) 트리 크기 찾기
best_tree_size = min(scores, key=scores.get)
```

### [2단계] : 최적의 모델로 전체 데이터 학습 (Step 2: Fit Model Using All Data)
이제 최적의 트리 크기(best_tree_size)를 사용하여 전체 데이터로 모델을 학습합니다. 검증 데이터를 더 이상 유지할 필요 없이, 전체 데이터(X, y)를 사용하여 모델을 훈련합니다.

```python
# 최적의 트리 크기를 적용한 모델 생성
final_model = DecisionTreeRegressor(max_leaf_nodes=best_tree_size, random_state=1)

# 모델 학습
final_model.fit(X, y)
```

결정 트리(Decision Tree)는 깊이와 정확도 사이에서 균형을 맞추는 것이 어려운 모델입니다.

깊은 트리(Deep Tree, 많은 리프 노드)
- 너무 구체적인 예측을 수행하여 과적합(Overfitting) 될 가능성이 높습니다.
- 각 리프(leaf)에는 소수의 집 데이터만 포함되므로, 새로운 데이터에 대한 예측이 불안정할 수 있습니다.

얕은 트리(Shallow Tree, 적은 리프 노드)
- 데이터의 다양한 패턴을 학습하지 못해 과소적합(Underfitting) 될 가능성이 높습니다.
- 주어진 데이터의 차이를 충분히 구분하지 못해 성능이 낮습니다.

이러한 과적합과 과소적합 문제는 더 정교한 모델에서도 발생하는 공통적인 문제입니다.
하지만 일부 모델들은 스마트한 방법으로 성능을 향상시킵니다.
이번에는 랜덤 포레스트(Random Forest) 를 예제로 살펴보겠습니다.

### 랜덤 포레스트란?
- 랜덤 포레스트는 **여러 개의 결정 트리를 조합하여 예측을 수행하는 모델**입니다.
- 각 개별 트리의 예측을 평균 내어 최종 결과를 생성합니다.
- 단일 결정 트리보다 춸씬 더 좋은 예측 성능을 보여줍니다.
- **기본 설정만으로도 우수한 성능**을 발휘합니다.
- 추가적인 하이퍼파라미터 조정으로 성능을 더욱 개선할 수 있습니다.

### 랜덤 포레스트 모델 학습시켜보기 

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 랜덤 포레스트 모델 생성
forest_model = RandomForestRegressor(random_state=1)

# 모델 학습
forest_model.fit(train_X, train_y)

# 검증 데이터 예측
melb_preds = forest_model.predict(val_X)

# MAE(평균 절대 오차) 출력
print(mean_absolute_error(val_y, melb_preds))
```

물론, 랜덤 포레스트의 성능을 더 개선하기 위해 다양한 하이퍼파라미터 조정이 가능합니다.
하지만 기본 설정만으로도 충분히 우수한 성능을 발휘하는 것이 랜덤 포레스트의 가장 큰 장점입니다.
