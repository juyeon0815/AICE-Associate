### XGBoost
- 모델을 앙상블에 반복적으로 추가하기 위해 사이클을 거치는 방법

1. 현재 앙상블을 사용하여 데이터 세트의 각 관측치에 대한 에측 생성
2. 예측을 실제 값과 비교
3. 예측을 개선하기 위해 새로운 모델을 구축하는 데 이러한 비교에서 정보 사용

→ 1단계와 2단계는 데이터를 통해 예측하고 예측을 실제 값과 비교하여 오류를 계산하는 것으로 나타낸다. 3단계는 오류 패턴에 따라 예측 변수를 조정하여 새로운 예측 변수를 구축하는 것. 이후, 새로운 예측 모델을 이전 모델과 결합하여 앙상블의 예측을 개선
